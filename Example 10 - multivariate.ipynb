## Model

$$Correct_i \sim Bernoulli(p_i)$$
$$RT_i \sim LogNormal(\mu_i, \sigma_{RT})$$
$$logit(p_i) \sim \alpha_{Correct}[P_i] + \beta_{Correct}[P_i]\cdot S_i$$
$$log(mu_i) \sim \alpha_{RT}[P_i] + \beta_{RT}[P_i] \cdot S_i$$

$$\begin{aligned}
\begin{bmatrix}
\alpha_{Correct}[P_i] \\
\alpha_{RT}[P_i] \\
\beta_{Correct}[P_i] \\
\beta_{RT}[P_i]
\end{bmatrix} 
&\sim \text{MVNormal}\left( 
\begin{bmatrix}
\alpha_{\text{Correct}} \\
\alpha_{\text{RT}} \\
\beta_{\text{Correct}} \\
\beta_{\text{RT}}
\end{bmatrix}, \mathbf{S} 
\right) \\[10pt]
\mathbf{S} &= 
\begin{pmatrix}
\sigma_{\alpha_{\text{Correct}}} & 0 & 0 & 0 \\
0 & \sigma_{\alpha_{\text{RT}}} & 0 & 0 \\
0 & 0 & \sigma_{\beta_{\text{Correct}}} & 0 \\
0 & 0 & 0 & \sigma_{\beta_{\text{RT}}}
\end{pmatrix} 
\mathbf{R} 
\begin{pmatrix}
\sigma_{\alpha_{\text{Correct}}} & 0 & 0 & 0 \\
0 & \sigma_{\alpha_{\text{RT}}} & 0 & 0 \\
0 & 0 & \sigma_{\beta_{\text{Correct}}} & 0 \\
0 & 0 & 0 & \sigma_{\beta_{\text{RT}}}
\end{pmatrix}
\end{aligned}
$$

$$\alpha_{Correct} \sim Normal(logit(0.75), 1)$$
$$\alpha_{RT} \sim Normal(log(1), 1)$$
$$\beta_{Correct} \sim Normal(0, 1)$$
$$\beta_{RT} \sim Normal(0, 1)$$
$$\sigma_{\alpha_{\text{Correct}}}, 
   \sigma_{\alpha_{\text{RT}}},
   \sigma_{\beta_{\text{Correct}}}, 
   \sigma_{\alpha_{\text{RT}}} \sim Exponential(1)$$

$$R \sim LKJcorr(2)$$
$$\sigma_{RT} \sim Exponential(1)$$
