## Model

$$D_i \sim Gamma(\sigma_i, \rho)$$
$$log(\sigma_i) = \begin{cases}
\alpha + \beta \cdot H_i & \text{for } S_i = 1\\
\alpha + \beta \cdot (1 - H_i) & \text{for } S_i = 0
\end{cases}$$

$$H_i = \begin{cases} 
      H_{start} & \text{for } i = 1 \\
      H_{i-1}  + (H_{i-1} - S_{i-1}) e^{-\frac{D_{i-1}}{\tau_H}} & \text{for } i > 1 
   \end{cases}$$
   
$$\alpha \sim Normal(log(2), 0.5)$$
$$beta \sim Normal(0, 1)$$
$$H_{start} \sim Beta(3, 3)$$
$$\tau_H \sim Exponential(1)$$
$$\rho \sim Exponential(1)$$
