
In psychophysical experiment, e.g., a direction discrimination task pictured below, participants' responses may reflect influence of prior experience in a phenomenon dubbed "serial dependence".

![](images/example07-experiment.png)

We can build a model that incorporates this influence based on overall importance of prior information ($W_{prev}$) and its relevance during the current trial based on Gaussian distance between prior respones and current perceptual evidence controlled by parameter $\lambda$.

## Model

$$R_i \sim Normal(\mu_i, \sigma)$$
$$\mu_i = \begin{cases} 
      D_i & \text{for } i = 1 \\
      D_i  + W_{prev} \cdot e^{-\frac{(R_{i-1}-D_i)^2}{2\lambda^2}} \cdot (R_{i-1}-D_i) & \text{for } i > 1 
   \end{cases}$$
$$W_{prev} \sim Normal(0, 1)$$ 
$$\lambda \sim Exponential(1)$$
$$\sigma \sim Exponential(1)$$
