## Model

Full effect model

$$O_i \sim Poisson(\lambda^F_i)$$
$$log(\lambda^F_i) = \alpha^F + \beta^F_P\cdot P_i$$
$$\alpha^F \sim Normal(log(12), 1)$$
$$\beta^F_P \sim Normal(0, 1)$$

Direct effect model

$$O_i \sim Poisson(\lambda^D_i)$$
$$log(\lambda^D_i) = \alpha^D + \beta^D_P\cdot P_i + \beta^D_M\cdot M_i$$
$$\alpha^D \sim Normal(log(12), 1)$$
$$\beta^D_P \sim Normal(0, 1)$$
$$\beta^D_M \sim Normal(0, 1)$$

Mediated effect
$$\beta^M_P = \beta^F_P - \beta^D_P$$
